```python
!pip install -q tensorflow tf2onnx

import tensorflow as tf
import numpy as np
import tf2onnx

# -------------------------
# MODEL (CLEAN FUNCTIONAL API)
# -------------------------
inputs = tf.keras.Input(shape=(128,128,3), name="input")

x = tf.keras.layers.Conv2D(16, 3, activation='relu')(inputs)
x = tf.keras.layers.MaxPooling2D()(x)
x = tf.keras.layers.Conv2D(32, 3, activation='relu')(x)
x = tf.keras.layers.MaxPooling2D()(x)
x = tf.keras.layers.Flatten()(x)
outputs = tf.keras.layers.Dense(4, activation='softmax', name="output")(x)

model = tf.keras.Model(inputs, outputs)

# -------------------------
# TRAIN (FAKE BUT VALID)
# -------------------------
X = np.random.rand(100,128,128,3).astype(np.float32)
y = np.random.randint(0,4,100)

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy')
model.fit(X, y, epochs=1, verbose=0)

# -------------------------
# FIXED EXPORT (STABLE PATH)
# -------------------------
spec = (tf.TensorSpec((None,128,128,3), tf.float32, name="input"),)

onnx_model, _ = tf2onnx.convert.from_keras(
    model,
    input_signature=spec,
    opset=13
)

# -------------------------
# SAVE
# -------------------------
with open("waste_model.onnx", "wb") as f:
    f.write(onnx_model.SerializeToString())

print("✅ ONNX MODEL READY FOR STREAMLIT")
```
